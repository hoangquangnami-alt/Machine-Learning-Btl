# 03 · Đặc trưng

Bài toán dùng 18 đặc trưng được chọn cho thời điểm nhập học; một số thời điểm ghi nhận vẫn cần xác minh.

Notebook chạy độc lập từ thư mục gốc hoặc `notebooks/`; cài `requirements-demo.txt` trước.

```python
from pathlib import Path
import sys, json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
ROOT = Path.cwd().resolve()
if not (ROOT/'student_success').exists(): ROOT = ROOT.parent
assert (ROOT/'student_success').exists(), 'Chạy từ thư mục dự án hoặc notebooks/'
sys.path.insert(0, str(ROOT))
from student_success.data import CATEGORICAL, NUMERIC, LABELS, SEED, split
DATA = ROOT/'data/raw/data.csv'
assert DATA.exists(), 'Chạy python scripts/download_data.py'
df = pd.read_csv(DATA, sep=';'); df.columns = df.columns.str.strip()
train_idx, test_idx = split(df)
train = df.loc[train_idx]
R = ROOT/'data/experiment'
```

```python
from student_success.data import EXCLUDED
print('Số:', len(NUMERIC), '| Phân loại:', len(CATEGORICAL), '| Không dùng:', len(EXCLUDED))
display(pd.DataFrame({'Đặc trưng': NUMERIC+CATEGORICAL, 'Loại': ['Số']*len(NUMERIC)+['Phân loại']*len(CATEGORICAL)}))
print('Các trường không dùng:', ', '.join(EXCLUDED))
```

```python
from student_success.preprocess import build_preprocessor
pre = build_preprocessor('A')
X_train = pre.fit_transform(train[CATEGORICAL+NUMERIC])
print('Dạng ma trận sau fit trên train:', X_train.shape)
print('Ví dụ tên cột:', pre.get_feature_names_out()[:15])
```
